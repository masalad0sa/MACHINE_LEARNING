```python
from sklearn.preprocessing import OneHotEncoder
X = df[['Payment Method']]
encoder= OneHotEncoder(drop='first', sparse_output=False)
encoder_df = pd.DataFrame(encoder.fit_transform(X))
encoder_df
```

```python
encoded_features = encoder.fit_transform(X)

# 5. Convert the result back to a DataFrame and join with original data
encoder_df = pd.DataFrame(encoded_features, columns=encoder.get_feature_names_out(['Payment Method']))
df = pd.concat([df.reset_index(drop=True), encoder_df], axis=1)

# Display the result
df
```
