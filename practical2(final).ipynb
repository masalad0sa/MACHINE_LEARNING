```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

# Creating a sample dataset
data = {
    "Marks": [78, 85, np.nan, 92, 88, 95, 60],
    "Study_Hours": [4, 6, 5, np.nan, 7, 8, 3],
    "Result": ["Pass", "Pass", "Pass", "Pass", "Pass", np.nan, "Fail"],
    "Branch": ["IT", "CSE", "IT", "ECE", "CSE", "IT", "ECE"]
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)

```

```python
# Handling missing values
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].median())
df["Result"] = df["Result"].fillna(df["Result"].mode()[0])

print("\nAfter Handling Missing Values:\n", df)
```

```python

# Label Encoding
le = LabelEncoder()
df["Result_Label"] = le.fit_transform(df["Result"])

print("\nAfter Label Encoding:\n", df)
```

```python

# One-Hot Encoding
df = pd.get_dummies(df, columns=["Branch"])

print("\nAfter One-Hot Encoding:\n", df)

```

```python
# Normalization
scaler = MinMaxScaler()
df[["Marks", "Study_Hours"]] = scaler.fit_transform(df[["Marks", "Study_Hours"]])

print("\nAfter Normalization:\n", df)

```

```python
# Standardization
std = StandardScaler()
df[["Marks", "Study_Hours"]] = std.fit_transform(df[["Marks", "Study_Hours"]])

print("\nAfter Standardization:\n", df)

```

```python
# Statistical Measures
mean_val = df["Marks"].mean()
median_val = df["Marks"].median()
mode_val = df["Marks"].mode()[0]
variance_val = df["Marks"].var()
std_dev_val = df["Marks"].std()

Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)
IQR = Q3 - Q1

print("\nSTATISTICAL MEASURES (Marks Column)")
print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
print("Variance:", variance_val)
print("Standard Deviation:", std_dev_val)
print("Q1:", Q1)
print("Q3:", Q3)
print("Interquartile Range (IQR):", IQR)

```

```python
# Outlier Detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["Marks"] < lower_bound) | (df["Marks"] > upper_bound)]

print("\nOutliers Detected Using IQR:\n", outliers)
print("\nFinal Preprocessed Dataset:\n", df)
```
